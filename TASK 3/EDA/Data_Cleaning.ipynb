### DATA CLEANING AND PREPARATION

```python
# Checking for missing values in the dataset
missing_values = health_data.isnull().sum()
missing_values[missing_values > 0]
```

```python
# checking for duplicate values in the dataset
duplicate_values = health_data.duplicated().sum()
duplicate_values
```

```python
# Removing duplicate values from the dataset
health_data_cleaned = health_data.drop_duplicates()
health_data_cleaned.shape
```

```python
# New Summary statistics of the dataset after removing duplicates
health_data_cleaned.describe().T
```

```python
# saving the cleaned dataset to a new excel file
health_data_cleaned.to_excel("../DATASETS/healthcare_dataset_cleaned.xlsx", index=False)
```
